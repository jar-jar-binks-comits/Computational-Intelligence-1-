# EVOLUTIONARY STRATEGIES

### What is Optimisation? 
Finding a vector *x* that minimises an objective function *f(x)*
$$\min_{x \in \mathbb{R}^n} f(x)$$
A solution *x* is a **global minimum** if no other point in the entire search produces a lower value:
$$f(x^*) \leq f(x) \quad \forall x \in \mathbb{R}^n$$
A **local minimum** is only better than its immediate neighbours. 

### What is Black-Box Optimisation?
Where a function has no known formula, no accessible derivatives, and can only be evaluated by running it. 
So the **gradient-based method** fials here because they require derivate info.
EA are built for these settings since they only need to evaluate solutions, not understand thier internal structure 

### Evolutionary Algorithms (EA)
EA is population-based optimisation method, where each individual is a candidate solution encode as **genotype** (internal representation, *e.g., real-valued vector*) mapped to a **phenotype** (actual solution in the problem space).
> **fitness function** which scores solution quality
> 
> Over generations, the population evolves through:
> 1. **mutation** $\rightarrow$ random perturbation of a solution
> 2. **recombination** $\rightarrow$ combining info from 2 solutions 
> 3. **selection** $\rightarrow$ keeping the best 

### Evolutionary Strategies (ES)
ES specific class of EA designed for continuous black-box optimisation, and it perates direclty on *real-valued vectors* and *adapt* their mutation behaviour based on *search feedback*. 
- so the difference with *genetic algorithm* (GA) is that GA uses binary/symbolic representations
